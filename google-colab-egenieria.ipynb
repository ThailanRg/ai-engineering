# Integrando com API Gemini

```python
import os
from google.colab import userdata

os.environ["GOOGLE_API_KEY"] = userdata.get('GEMINI_API_KEY')
```

```python
from google import genai
client = genai.Client()
```

```python
chat = client.chats.create(model = "gemini-2.5-flash")
```

```python
promp = input("Digite uma pergunta: ")

while promp != "fim":
  try:
    resposta = chat.send_message(promp)
    print(resposta.text)
  except Exception as e:
    print("Limite atingido, aguardando...")
  promp = input("Digite uma pergunta: ")
```

# Estruturas de repetição

# Lista

```python
nome_alunos = ["João Silva", "Maria Souza", "Carlos Oliveira", "Ana Pereira", "Lucas Santos"]
print(nome_alunos)

print(len(nome_alunos))

n = 0
while n < len(nome_alunos):
  print(nome_alunos[n])
  n = n + 1
```
