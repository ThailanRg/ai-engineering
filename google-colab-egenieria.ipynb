# Integrando com API Gemini

```python
import os
from google.colab import userdata

os.environ["GOOGLE_API_KEY"] = userdata.get('GEMINI_API_KEY')
```

```python
from google import genai
client = genai.Client()
```

```python
chat = client.chats.create(model = "gemini-2.5-flash")
```

```python
promp = input("Digite uma pergunta: ")

while promp != "fim":
  try:
    resposta = chat.send_message(promp)
    print(resposta.text)
  except Exception as e:
    print("Limite atingido, aguardando...")
  promp = input("Digite uma pergunta: ")
```

# Lista

```python
nome_alunos = ["João Silva", "Maria Souza", "Carlos Oliveira", "Ana Pereira", "Lucas Santos"]
print("------------ imprimir lista ------------ ")
print("")
print(nome_alunos)
print("")
print("------------ imprimir a quantidade de itens ------------ ")
print("")
print(len(nome_alunos))
print("")
print("------------ imprimir item na posicao 0 ao 2 ------------ ")
print("")
print(nome_alunos[0:2])
print("")
print("------------ imprimir da posicao 2 ate o final da lista ------------ ")
print("")
print(nome_alunos[1:])
print("")
print("------------ imprimir da posicao 2 ate o primeiro ------------ ")
print("")
print(nome_alunos[:2])
print("")
print("------------ adicionar item na lista ------------ ")
print("")
nome_alunos.append("Pedro Santos")
print(nome_alunos)
print("")
print("------------ adicionar multiplos itens ------------ ")
print("")
nome_alunos.extend(["a", "b", "c"])
print(nome_alunos)
print("")
print("------------ remover item ------------ ")
print("")
nome_alunos.remove("a")
nome_alunos.pop(len(nome_alunos)-1)
nome_alunos.pop()
print(nome_alunos)
print("")
print("------------ indice negativo ------------ ")
print("")
print(nome_alunos[-1])
print("")
print("------------ iterar pela lista e imprimir ------------ ")
print("")

n = 0
while n < len(nome_alunos):
  print(nome_alunos[n])
  n = n + 1
```

```python
media_alunos = [7.5, 8.0, 6.8, 9.2, 7.9]
print(len(media_alunos))

n = 0
while n < len(media_alunos):
  if(media_alunos[n] >= 7):
    print(media_alunos[n])
  n = n + 1
```

```python
nome_1 = "   jOÃo    sIlVa   "
nome_2 = "mArIa     doS   sAnToS"
nome_3 = "   peDro    ALMEIDA"
nome_4 = "aNa    cLaRa   ferReIrA   "
nome_5 = "   luCaS    RoDrIgUeS   "
nome_6 = "CaMiLa    dE   oLiVeIrA"
nome_7 = "   rAfAeL   coStA"
nome_8 = "beAtRiZ     soUzA   "
nome_9 = "   feLiPe    lImA   "
nome_10 = "jÚlIa    baRBoSa    "



def fomatString(nome):
  maiusculo = nome.upper()
  removerEspaco = maiusculo.replace("  ", " ")
  removerEspaco2 = removerEspaco.strip()
  listaPalabras = removerEspaco2.split()
  format = " ".join(listaPalabras)
  return format

print(fomatString(nome_1))
print(fomatString(nome_2))
print(fomatString(nome_3))
```

```python
import random

nome_11 = "Ana Julia"
salas_de_aulas = ["sala1", "sala2", "sala3"]

def aloca_alunos_em_salas(nome_do_aluno, lista_de_salas):
    sala_escolhida = random.choice(lista_de_salas)
    dict_aluno = {
        "nome": nome_do_aluno,
        "sala": sala_escolhida
    }
    return dict_aluno

print(aloca_alunos_em_salas(nome_11, salas_de_aulas)    )
```
